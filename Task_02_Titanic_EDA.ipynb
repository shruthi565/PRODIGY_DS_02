# Task-02: Data Cleaning and Exploratory Data Analysis on Titanic Dataset

## Introduction

The Titanic dataset contains information about passengers who travelled on the RMS Titanic. It includes details such as passenger age, gender, passenger class, fare, number of siblings or spouses, number of parents or children, and survival status.

In this task, data cleaning and exploratory data analysis (EDA) are performed to understand the dataset, identify missing values and patterns, and explore relationships between different variables and passenger survival.

Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn are used for data processing and visualization.



## Objective

The main objectives of this task are:

* To load and understand the Titanic dataset.
* To identify and handle missing values.
* To check and remove duplicate records.
* To perform basic data cleaning.
* To analyze the distribution of important variables.
* To study the relationship between passenger characteristics and survival.
* To visualize patterns and trends using graphs.
* To identify the factors that influenced passenger survival.


Import Required Libraries

```python
url= "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df=pd.read_csv(url)
print("Dataset loaded successfully")
```

#Data Cleaning

Checking Missing Value

```python
df.isnull().sum()
```

Visualizing Missing Value

```python
plt.figure(figsize=(10,5))
sns.heatmap(df.isnull(),cbar=False)
plt.title("Missing Values in Titanic Dataset")
plt.show()
```

Checking duplicate values

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

Handle Missing Age

```python
df['Age']=df['Age'].fillna(df['Age'].median())
print("Missing Age values:", df["Age"].isnull().sum())
```

Handle Cabin Column

```python
df['Cabin']=df['Cabin'].fillna(df['Cabin'].mode()[0])
print("Missing Cabin values:", df['Cabin'].isnull().sum())
```

Handle Embarked

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
print("Missing Embarked values:",df['Embarked'].isnull().sum())
```

Checking Missing Values again

```python
print(df.isnull().sum())
```

Checking datatypes

```python
df.dtypes
```

Convert Categorical Variables

```python
df['Sex']=df['Sex'].astype('category')
df['Embarked']=df['Embarked'].astype('category')
```

```python
df.dtypes
```

Checking Cleaned Dataset




```python
df.head()
```

```python
df.tail()
```

#Exploratory Data Analysis

Survival Count

```python
plt.figure(figsize=(6, 5))
sns.countplot(x="Survived", data=df)
plt.title("Passenger Survival Count")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

Survival Percentage

```python
survival_percentage=df['Survived'].value_counts(normalize=True)*100
print(survival_percentage)
```

```python
plt.figure(figsize=(6, 5))
plt.pie(survival_percentage,labels=["Did Not Survive", "Survived"],autopct="%1.1f%%",startangle=90)
plt.title("Passenger Survival Percentage")
plt.show()
```

Survival by Gender


```python
plt.figure(figsize=(7,5))
sns.countplot(x='Sex',hue='Survived',data=df)
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

Calculate Survival Rate by Gender

```python
gender_survival = df.groupby("Sex", observed=True)["Survived"].mean() * 100
print(gender_survival)
```

```python
plt.figure(figsize=(7, 5))
gender_survival.plot(kind="bar")
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

Survival by Passenger Class


```python
class_survival = df.groupby("Pclass")["Survived"].mean() * 100
print(class_survival)
```

```python
plt.figure(figsize=(7, 5))
class_survival.plot(kind="bar")
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

Age Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Age"], bins=30, kde=True)
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

Age vs Survival

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x="Survived", y="Age", data=df)
plt.title("Age Distribution by Survival")
plt.xlabel("Survived")
plt.ylabel("Age")
plt.show()
```

Fare Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Fare"], bins=30, kde=True)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```

Fare vs Survival

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="Survived", y="Fare", data=df)
plt.title("Fare Distribution by Survival")
plt.xlabel("Survived")
plt.ylabel("Fare")
plt.show()
```

Survival by Embarked Location

```python
plt.figure(figsize=(7, 5))
sns.countplot(x="Embarked", hue="Survived", data=df)
plt.title("Survival by Embarkation Port")
plt.xlabel("Port")
plt.ylabel("Number of Passengers")
plt.show()
```

Survival Rate by Embarked Port

```python
embarked_survival = df.groupby("Embarked", observed=True)["Survived"].mean() * 100
print(embarked_survival)
```

```python
plt.figure(figsize=(7, 5))
embarked_survival.plot(kind="bar")
plt.title("Survival Rate by Embarkation Port")
plt.xlabel("Embarked Port")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

Number of Siblings/Spouses

```python
plt.figure(figsize=(8, 5))
sns.countplot(x="SibSp", data=df)
plt.title("Number of Siblings/Spouses Aboard")
plt.xlabel("SibSp")
plt.ylabel("Number of Passengers")
plt.show()
```

Number of Parents/Children

```python
plt.figure(figsize=(8, 5))
sns.countplot(x="Parch", data=df)
plt.title("Number of Parents/Children Aboard")
plt.xlabel("Parch")
plt.ylabel("Number of Passengers")
plt.show()
```

Family size


```python
df['FamilySize']=df['SibSp']+df['Parch']+1
df[["SibSp", "Parch", "FamilySize"]].head()
```

Analyze Family Size vs Survival

```python
plt.figure(figsize=(10, 5))
sns.countplot(x="FamilySize", hue="Survived", data=df)
plt.title("Family Size vs Survival")
plt.xlabel("Family Size")
plt.ylabel("Number of Passengers")
plt.show()
```

Create Alone Feature

```python
df["IsAlone"] = np.where(df["FamilySize"] == 1, 1, 0)
df[["FamilySize", "IsAlone"]].head()
```

Survival by Alone Status

```python
plt.figure(figsize=(7, 5))
sns.countplot(x="IsAlone", hue="Survived", data=df)
plt.title("Survival by Alone Status")
plt.xlabel("Is Alone")
plt.ylabel("Number of Passengers")
plt.show()
```

Survival Rate of Alone vs Not Alone

```python
alone_survival = df.groupby("IsAlone")["Survived"].mean() * 100
print(alone_survival)
```

Gender + Passenger Class

```python
plt.figure(figsize=(8, 5))
sns.barplot(x="Pclass",y="Survived",hue="Sex",data=df)
plt.title("Survival Rate by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

Age + Gender + Survival

```python
plt.figure(figsize=(9, 6))
sns.displot(data=df, x="Age", hue="Survived", col="Sex", bins=20, kde=True)
plt.show()
```

Correlation Matrix

```python
numeric_df = df.select_dtypes(include=np.number)
correlation = numeric_df.corr()
print(correlation)
```

Correlation Heatmap


```python
plt.figure(figsize=(10, 7))
sns.heatmap(correlation,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

Survival Rate Table

```python
print("Survival Rate by Gender:")
print(df.groupby("Sex", observed=True)["Survived"].mean())

print("\nSurvival Rate by Passenger Class:")
print(df.groupby("Pclass")["Survived"].mean())

print("\nSurvival Rate by Embarked Port:")
print(df.groupby("Embarked", observed=True)["Survived"].mean())
```

Pivot Table

```python
pivot_table = pd.pivot_table(df,values="Survived",index="Sex",columns="Pclass",aggfunc="mean")
print(pivot_table)
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(pivot_table,annot=True,fmt=".2f",cmap="YlGnBu")
plt.title("Survival Rate by Gender and Passenger Class")
plt.show()
```
